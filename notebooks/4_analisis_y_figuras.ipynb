```python
# =====================================================
# Análisis de Latencia por LLM - Código Python para Google Colab
# Autor: Tu nombre - Tesis de Investigación
# Fecha: Agosto 2025
# =====================================================

# =====================================================
# 1. INSTALACIÓN DE LIBRERÍAS Y CONFIGURACIÓN INICIAL
# =====================================================

# Ya instalado en la primera celda

# =====================================================
# 2. CARGA Y PREPARACIÓN DE DATOS
# =====================================================

# Ya cargado en la segunda celda

# =====================================================
# 3. CONFIGURACIÓN DE COLORES Y ETIQUETAS CONSISTENTES
# =====================================================

# Definir LLMs y sus colores consistentes
llms = ['gpt4o_mini', 'gemini2_flash', 'ministral_8b', 'deepseek_r1', 'llama33_70b']
llm_labels = {
    'gpt4o_mini': 'GPT-4o Mini',
    'gemini2_flash': 'Gemini-2 Flash',
    'ministral_8b': 'Ministral-8B',
    'deepseek_r1': 'DeepSeek-R1',
    'llama33_70b': 'Llama-3.3-70B'
}

# Paleta de colores consistente para todos los gráficos
llm_colors = {
    'GPT-4o Mini': '#8884d8',      # Azul
    'Gemini-2 Flash': '#82ca9d',   # Verde
    'Ministral-8B': '#ffc658',     # Amarillo/Naranja
    'DeepSeek-R1': '#ff7c7c',      # Rojo/Rosa
    'Llama-3.3-70B': '#8dd1e1'     # Cyan
}

# Colores para dificultad
difficulty_colors = {
    'easy': '#82ca9d',    # Verde
    'medium': '#ffc658',  # Amarillo
    'hard': '#ff7c7c'     # Rojo
}

# =====================================================
# 4. PREPARACIÓN DE DATOS PARA ANÁLISIS
# =====================================================

def prepare_latency_data(df):
    """Prepara los datos de latencia en formato largo para análisis"""
    latency_data = []

    for _, row in df.iterrows():
        for llm in llms:
            latency = row.get(f'latency_{llm}')
            response_length = row.get(f'response_length_{llm}')

            if pd.notna(latency) and latency > 0:
                latency_data.append({
                    'llm': llm_labels[llm],
                    'llm_short': llm,
                    'difficulty': row['difficulty'],
                    'latency': float(latency),
                    'response_length': float(response_length) if pd.notna(response_length) else np.nan,
                    'paper_id': row['paper_id'],
                    'question_id': row['question_id'],
                    'question_type': row.get('question_type', 'unknown'),
                    'domain': row.get('domain_detected', 'unknown')
                })

    return pd.DataFrame(latency_data)

# Preparar datos
latency_df = prepare_latency_data(df)
print(f"Datos de latencia preparados: {len(latency_df)} registros")

# Calcular eficiencia (latencia/longitud de respuesta)
latency_df['efficiency'] = latency_df['latency'] / latency_df['response_length']
latency_df['efficiency'] = latency_df['efficiency'].replace([np.inf, -np.inf], np.nan)

# =====================================================
# 5. GRÁFICO 1: BARRAS AGRUPADAS - LATENCIA PROMEDIO
# =====================================================

def plot_grouped_bar_chart(data):
    """Gráfico de barras agrupadas por LLM y dificultad"""
    # Calcular estadísticas
    stats_df = data.groupby(['llm', 'difficulty'])['latency'].agg(['mean', 'std', 'count']).reset_index()

    # Crear pivot para el gráfico
    pivot_data = stats_df.pivot(index='difficulty', columns='llm', values='mean')
    pivot_std = stats_df.pivot(index='difficulty', columns='llm', values='std')

    # Ordenar dificultades
    difficulty_order = ['easy', 'medium', 'hard']
    pivot_data = pivot_data.reindex(difficulty_order)
    pivot_std = pivot_std.reindex(difficulty_order)

    # Crear gráfico
    fig, ax = plt.subplots(figsize=(14, 8))

    x = np.arange(len(difficulty_order))
    width = 0.15

    for i, llm in enumerate(pivot_data.columns):
        values = pivot_data[llm]
        errors = pivot_std[llm]
        pos = x + (i - 2) * width

        ax.bar(pos, values, width, label=llm, color=llm_colors[llm],
               yerr=errors, capsize=5, alpha=0.8)

        # Añadir valores en las barras
        for j, v in enumerate(values):
            ax.text(pos[j], v + errors.iloc[j] + 0.1, f'{v:.2f}s',
                   ha='center', va='bottom', fontsize=10)

    ax.set_xlabel('Dificultad de la Pregunta', fontsize=14, fontweight='bold')
    ax.set_ylabel('Latencia Promedio (segundos)', fontsize=14, fontweight='bold')
    ax.set_title('Latencia Promedio por LLM y Dificultad de Pregunta\n(con desviación estándar)',
                fontsize=16, fontweight='bold', pad=20)
    ax.set_xticks(x)
    ax.set_xticklabels([d.capitalize() for d in difficulty_order])
    ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    return stats_df

stats_summary = plot_grouped_bar_chart(latency_df)
print("\nEstadísticas por LLM y Dificultad:")
print(stats_summary.round(3))

# =====================================================
# 6. GRÁFICO 2: HEATMAP DE LATENCIA
# =====================================================

def plot_latency_heatmap(data):
    """Heatmap de latencia promedio por LLM vs Dificultad"""
    # Preparar datos para heatmap
    heatmap_data = data.groupby(['llm', 'difficulty'])['latency'].mean().reset_index()
    pivot_heatmap = heatmap_data.pivot(index='llm', columns='difficulty', values='latency')

    # Ordenar filas y columnas
    difficulty_order = ['easy', 'medium', 'hard']
    llm_order = ['GPT-4o Mini', 'Gemini-2 Flash', 'Ministral-8B', 'DeepSeek-R1', 'Llama-3.3-70B']

    pivot_heatmap = pivot_heatmap.reindex(index=llm_order, columns=difficulty_order)

    # Crear heatmap
    fig, ax = plt.subplots(figsize=(10, 8))

    sns.heatmap(pivot_heatmap, annot=True, fmt='.2f', cmap='RdYlBu_r',
                center=pivot_heatmap.values.mean(), square=True,
                linewidths=0.5, cbar_kws={"shrink": .8}, ax=ax)

    ax.set_title('Heatmap: Latencia Promedio por LLM y Dificultad\n(segundos)',
                fontsize=16, fontweight='bold', pad=20)
    ax.set_xlabel('Dificultad de la Pregunta', fontsize=14, fontweight='bold')
    ax.set_ylabel('Modelo de Lenguaje (LLM)', fontsize=14, fontweight='bold')
    ax.set_xticklabels([d.capitalize() for d in difficulty_order], rotation=0)
    ax.set_yticklabels(llm_order, rotation=0)

    plt.tight_layout()
    plt.show()

    return pivot_heatmap

heatmap_results = plot_latency_heatmap(latency_df)
print("\nMatriz de Latencia Promedio (segundos):")
print(heatmap_results.round(3))

# =====================================================
# 7. GRÁFICO 3: BOX PLOTS POR LLM
# =====================================================

def plot_boxplots_by_llm(data):
    """Box plots de distribución de latencia por LLM"""
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 12))

    # Box plot general por LLM
    llm_order = ['GPT-4o Mini', 'Gemini-2 Flash', 'Ministral-8B', 'DeepSeek-R1', 'Llama-3.3-70B']

    box_data = [data[data['llm'] == llm]['latency'].values for llm in llm_order]
    colors = [llm_colors[llm] for llm in llm_order]

    bp1 = ax1.boxplot(box_data, labels=llm_order, patch_artist=True,
                      showmeans=True, meanline=True)

    for patch, color in zip(bp1['boxes'], colors):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)

    ax1.set_title('Distribución de Latencia por LLM (Box Plots)',
                 fontsize=16, fontweight='bold')
    ax1.set_ylabel('Latencia (segundos)', fontsize=12, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.tick_params(axis='x', rotation=45)

    # Box plot por LLM y dificultad
    sns.boxplot(data=data, x='difficulty', y='latency', hue='llm',
                palette=llm_colors, ax=ax2)

    ax2.set_title('Distribución de Latencia por LLM y Dificultad',
                 fontsize=16, fontweight='bold')
    ax2.set_xlabel('Dificultad de la Pregunta', fontsize=12, fontweight='bold')
    ax2.set_ylabel('Latencia (segundos)', fontsize=12, fontweight='bold')
    ax2.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    ax2.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

plot_boxplots_by_llm(latency_df)

# =====================================================
# 8. GRÁFICO 4: VIOLIN PLOTS
# =====================================================

def plot_violin_plots(data):
    """Gráficas de violín para mostrar distribución de densidad"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

    # Violin plot por LLM
    llm_order = ['GPT-4o Mini', 'Gemini-2 Flash', 'Ministral-8B', 'DeepSeek-R1', 'Llama-3.3-70B']

    parts1 = ax1.violinplot([data[data['llm'] == llm]['latency'].values for llm in llm_order],
                           positions=range(len(llm_order)), showmeans=True, showmedians=True)

    for i, (pc, llm) in enumerate(zip(parts1['bodies'], llm_order)):
        pc.set_facecolor(llm_colors[llm])
        pc.set_alpha(0.7)

    ax1.set_xticks(range(len(llm_order)))
    ax1.set_xticklabels(llm_order, rotation=45, ha='right')
    ax1.set_ylabel('Latencia (segundos)', fontsize=12, fontweight='bold')
    ax1.set_title('Distribución de Densidad de Latencia por LLM\n(Violin Plots)',
                 fontsize=14, fontweight='bold')
    ax1.grid(True, alpha=0.3)

    # Violin plot por dificultad
    difficulty_order = ['easy', 'medium', 'hard']
    parts2 = ax2.violinplot([data[data['difficulty'] == diff]['latency'].values for diff in difficulty_order],
                           positions=range(len(difficulty_order)), showmeans=True, showmedians=True)

    for i, (pc, diff) in enumerate(zip(parts2['bodies'], difficulty_order)):
        pc.set_facecolor(difficulty_colors[diff])
        pc.set_alpha(0.7)

    ax2.set_xticks(range(len(difficulty_order)))
    ax2.set_xticklabels([d.capitalize() for d in difficulty_order])
    ax2.set_ylabel('Latencia (segundos)', fontsize=12, fontweight='bold')
    ax2.set_title('Distribución de Densidad de Latencia por Dificultad\n(Violin Plots)',
                 fontsize=14, fontweight='bold')
    ax2.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

plot_violin_plots(latency_df)

# =====================================================
# 9. ANÁLISIS DE OUTLIERS
# =====================================================

def analyze_outliers(data):
    """Identificación y análisis de valores atípicos"""
    def find_outliers_iqr(series):
        Q1 = series.quantile(0.25)
        Q3 = series.quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        return series[(series < lower_bound) | (series > upper_bound)]

    # Encontrar outliers por LLM
    outliers_summary = []

    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    axes = axes.flatten()

    llm_order = ['GPT-4o Mini', 'Gemini-2 Flash', 'Ministral-8B', 'DeepSeek-R1', 'Llama-3.3-70B']

    for i, llm in enumerate(llm_order):
        llm_data = data[data['llm'] == llm]['latency']
        outliers = find_outliers_iqr(llm_data)

        outliers_summary.append({
            'LLM': llm,
            'Total_Samples': len(llm_data),
            'Outliers_Count': len(outliers),
            'Outliers_Percentage': (len(outliers) / len(llm_data)) * 100,
            'Mean_Latency': llm_data.mean(),
            'Outliers_Mean': outliers.mean() if len(outliers) > 0 else np.nan,
            'Max_Outlier': outliers.max() if len(outliers) > 0 else np.nan
        })

        # Scatter plot con outliers destacados
        ax = axes[i]
        normal_data = llm_data[~llm_data.index.isin(outliers.index)]

        ax.scatter(range(len(normal_data)), sorted(normal_data),
                  alpha=0.6, color=llm_colors[llm], s=20, label='Normal')
        if len(outliers) > 0:
            ax.scatter(range(len(normal_data), len(normal_data) + len(outliers)),
                      sorted(outliers), color='red', s=40, alpha=0.8, label='Outliers')

        ax.set_title(f'{llm}\nOutliers: {len(outliers)} ({len(outliers)/len(llm_data)*100:.1f}%)')
        ax.set_xlabel('Índice de Muestra (ordenado)')
        ax.set_ylabel('Latencia (segundos)')
        ax.legend()
        ax.grid(True, alpha=0.3)

    # Ocultar el último subplot si no se usa
    if len(llm_order) < len(axes):
        axes[-1].set_visible(False)

    plt.suptitle('Análisis de Outliers por LLM\n(Método IQR: Q1-1.5*IQR, Q3+1.5*IQR)',
                fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()

    # Crear tabla resumen
    outliers_df = pd.DataFrame(outliers_summary)
    print("\nResumen de Outliers por LLM:")
    print(outliers_df.round(3))

    return outliers_df

outliers_analysis = analyze_outliers(latency_df)

# =====================================================
# 10. ANÁLISIS POR TIPO DE PREGUNTA
# =====================================================

def plot_performance_by_question_type(data):
    """Análisis de performance por tipo de pregunta"""
    # Filtrar tipos de pregunta con suficientes datos
    question_type_counts = data['question_type'].value_counts()
    valid_types = question_type_counts[question_type_counts >= 50].index

    filtered_data = data[data['question_type'].isin(valid_types)]

    # Calcular estadísticas por tipo de pregunta y LLM
    stats_by_type = filtered_data.groupby(['question_type', 'llm'])['latency'].agg(['mean', 'std', 'count']).reset_index()

    # Crear subplots
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 12))

    # Gráfico 1: Heatmap por tipo de pregunta
    heatmap_type = stats_by_type.pivot(index='llm', columns='question_type', values='mean')

    sns.heatmap(heatmap_type, annot=True, fmt='.2f', cmap='RdYlBu_r',
                center=heatmap_type.values.mean(), ax=ax1, cbar_kws={"shrink": .6})

    ax1.set_title('Latencia Promedio por LLM y Tipo de Pregunta',
                 fontsize=14, fontweight='bold')
    ax1.set_ylabel('Modelo de Lenguaje (LLM)', fontsize=12)
    ax1.set_xlabel('')

    # Gráfico 2: Box plot por tipo de pregunta
    sns.boxplot(data=filtered_data, x='question_type', y='latency', hue='llm',
                palette=llm_colors, ax=ax2)

    ax2.set_title('Distribución de Latencia por Tipo de Pregunta y LLM',
                 fontsize=14, fontweight='bold')
    ax2.set_xlabel('Tipo de Pregunta', fontsize=12, fontweight='bold')
    ax2.set_ylabel('Latencia (segundos)', fontsize=12, fontweight='bold')
    ax2.tick_params(axis='x', rotation=45)
    ax2.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    ax2.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    # Mostrar estadísticas
    print(f"\nTipos de pregunta analizados (≥50 muestras): {len(valid_types)}")
    print("Distribución por tipo de pregunta:")
    print(question_type_counts[valid_types].sort_values(ascending=False))

    return stats_by_type

question_type_analysis = plot_performance_by_question_type(latency_df)

# =====================================================
# 11. ANÁLISIS DE EFICIENCIA
# =====================================================

def plot_efficiency_analysis(data):
    """Análisis de eficiencia: ratio latencia/longitud de respuesta"""
    # Filtrar datos válidos (sin NaN en efficiency)
    valid_data = data.dropna(subset=['efficiency'])

    # Remover outliers extremos en eficiencia para mejor visualización
    Q1 = valid_data['efficiency'].quantile(0.25)
    Q3 = valid_data['efficiency'].quantile(0.75)
    IQR = Q3 - Q1
    filtered_data = valid_data[
        (valid_data['efficiency'] >= Q1 - 1.5 * IQR) &
        (valid_data['efficiency'] <= Q3 + 1.5 * IQR)
    ]

    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

    # 1. Scatter plot: Latencia vs Longitud de respuesta
    llm_order = ['GPT-4o Mini', 'Gemini-2 Flash', 'Ministral-8B', 'DeepSeek-R1', 'Llama-3.3-70B']

    for llm in llm_order:
        llm_data = filtered_data[filtered_data['llm'] == llm]
        ax1.scatter(llm_data['response_length'], llm_data['latency'],
                   alpha=0.6, color=llm_colors[llm], label=llm, s=20)

    ax1.set_xlabel('Longitud de Respuesta (caracteres)', fontsize=12)
    ax1.set_ylabel('Latencia (segundos)', fontsize=12)
    ax1.set_title('Latencia vs Longitud de Respuesta por LLM', fontsize=14, fontweight='bold')
    ax1.legend()
    ax1.grid(True, alpha=0.3)

    # 2. Box plot de eficiencia por LLM
    sns.boxplot(data=filtered_data, x='llm', y='efficiency',
                palette=llm_colors, ax=ax2)
    ax2.set_title('Eficiencia por LLM\n(Latencia/Longitud de Respuesta)',
                 fontsize=14, fontweight='bold')
    ax2.set_xlabel('Modelo de Lenguaje (LLM)', fontsize=12)
    ax2.set_ylabel('Eficiencia (s/char)', fontsize=12)
    ax2.tick_params(axis='x', rotation=45)
    ax2.grid(True, alpha=0.3)

    # 3. Eficiencia por dificultad
    sns.boxplot(data=filtered_data, x='difficulty', y='efficiency',
                palette=difficulty_colors, ax=ax3)
    ax3.set_title('Eficiencia por Dificultad de Pregunta', fontsize=14, fontweight='bold')
    ax3.set_xlabel('Dificultad', fontsize=12)
    ax3.set_ylabel('Eficiencia (s/char)', fontsize=12)
    ax3.grid(True, alpha=0.3)

    # 4. Heatmap de eficiencia promedio
    efficiency_heatmap = filtered_data.groupby(['llm', 'difficulty'])['efficiency'].mean().reset_index()
    efficiency_pivot = efficiency_heatmap.pivot(index='llm', columns='difficulty', values='efficiency')

    sns.heatmap(efficiency_pivot, annot=True, fmt='.4f', cmap='RdYlBu_r',
                center=efficiency_pivot.values.mean(), ax=ax4)
    ax4.set_title('Eficiencia Promedio por LLM y Dificultad', fontsize=14, fontweight='bold')
    ax4.set_ylabel('Modelo de Lenguaje (LLM)', fontsize=12)
    ax4.set_xlabel('Dificultad', fontsize=12)

    plt.tight_layout()
    plt.show()

    # Calcular estadísticas de eficiencia
    efficiency_stats = filtered_data.groupby('llm')['efficiency'].agg(['mean', 'median', 'std']).round(6)
    print("\nEstadísticas de Eficiencia por LLM (segundos/carácter):")
    print(efficiency_stats)

    return efficiency_stats

efficiency_analysis = plot_efficiency_analysis(latency_df)

# =====================================================
# 12. TABLA RESUMEN FINAL
# =====================================================

def generate_final_summary(data):
    """Genera tabla resumen final con todas las métricas"""

    summary_stats = []

    llm_order = ['GPT-4o Mini', 'Gemini-2 Flash', 'Ministral-8B', 'DeepSeek-R1', 'Llama-3.3-70B']

    for llm in llm_order:
        llm_data = data[data['llm'] == llm]

        summary_stats.append({
            'LLM': llm,
            'N_Samples': len(llm_data),
            'Mean_Latency_s': llm_data['latency'].mean(),
            'Median_Latency_s': llm_data['latency'].median(),
            'Std_Latency_s': llm_data['latency'].std(),
            'Min_Latency_s': llm_data['latency'].min(),
            'Max_Latency_s': llm_data['latency'].max(),
            'P95_Latency_s': llm_data['latency'].quantile(0.95),
            'P99_Latency_s': llm_data['latency'].quantile(0.99),
            'Mean_Response_Length': llm_data['response_length'].mean(),
            'Mean_Efficiency_s_per_char': llm_data['efficiency'].mean()
        })

    summary_df = pd.DataFrame(summary_stats)

    print("=" * 100)
    print("RESUMEN FINAL - ANÁLISIS DE LATENCIA POR LLM")
    print("=" * 100)
    print(summary_df.round(4))

    # Ranking por diferentes métricas
    print("\n" + "=" * 60)
    print("RANKINGS")
    print("=" * 60)

    print("\n1. RANKING POR LATENCIA PROMEDIO (mejor = menor):")
    ranking_latency = summary_df.sort_values('Mean_Latency_s')[['LLM', 'Mean_Latency_s']]
    for i, (_, row) in enumerate(ranking_latency.iterrows(), 1):
        print(f"   {i}. {row['LLM']}: {row['Mean_Latency_s']:.3f}s")

    print("\n2. RANKING POR EFICIENCIA (mejor = menor s/char):")
    efficiency_ranking = summary_df.dropna(subset=['Mean_Efficiency_s_per_char']).sort_values('Mean_Efficiency_s_per_char')
    for i, (_, row) in enumerate(efficiency_ranking.iterrows(), 1):
        print(f"   {i}. {row['LLM']}: {row['Mean_Efficiency_s_per_char']:.6f} s/char")

    print("\n3. RANKING POR CONSISTENCIA (mejor = menor desviación estándar):")
    consistency_ranking = summary_df.sort_values('Std_Latency_s')[['LLM', 'Std_Latency_s']]
    for i, (_, row) in enumerate(consistency_ranking.iterrows(), 1):
        print(f"   {i}. {row['LLM']}: σ = {row['Std_Latency_s']:.3f}s")

    return summary_df

final_summary = generate_final_summary(latency_df)
```
